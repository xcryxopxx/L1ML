### 4. Feature Engineering (Partea 1) - Argumentare și Idei

#### **A. Ce date externe ar putea îmbogăți setul de date?**
Pentru a crește acuratețea unei analize predictive sau a unui model de Machine Learning pe acest dataset (*Ames Housing Dataset*), setul de date ar putea fi îmbogățit cu următoarele date externe:
1. **Indicatori economici locali (Rata dobânzilor și inflația):** Rata dobânzilor la creditele ipotecare (*mortgage rates*) din anul și luna respectivă influențează direct puterea de cumpărare a cumpărătorilor și prețul final al locuințelor.
2. **Date geospațiale și de proximitate (*GIS*):** Distanța exactă în kilometri față de puncte de interes major din orașul Ames, cum ar fi: instituții de învățământ de top, parcuri, spitale și centre comerciale, precum și indicele de criminalitate din cartierul respectiv.
3. **Date despre eficiența energetică:** Informații actualizate privind costurile medii anuale de încălzire/răcire sau certificatul energetic al clădirii, aspecte care devin din ce în ce mai importante pentru cumpărători.

#### **B. Ce coloane existente sunt irelevante și pot fi eliminate?**
Anumite coloane din setul de date nu aduc nicio valoare predictivă reală pentru prețul unei case și pot fi eliminate pentru a reduce dimensionalitatea și zgomotul din date:
1. **Coloanele de tip ID / Index (`Order`, `PID`):** Acestea sunt doar numere de identificare a înregistrărilor sau a parcelelor în baza de date administrativă și nu au nicio corelație logică sau matematică cu valoarea de piață a locuinței.
2. **Coloane cu un grad extrem de ridicat de valori lipsă sau redundante:** De exemplu, coloane precum `Pool QC` (calitatea piscinei) sau `Misc Feature` (caracteristici diverse) au peste 95-99% valori lipsă (majoritatea caselor nu au piscină sau facilități speciale), ceea ce le face irelevante statistic, informația fiind deja acoperită parțial de coloana `Pool Area`.
